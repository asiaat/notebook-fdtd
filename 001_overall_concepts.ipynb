# Overview of concepts in vector analysis

## Scalar field

n mathematics and physics, a scalar field is a function that assigns a scalar value (i.e., specified by a single numerical quantity) to every point in space — for example, temperature, density, or air humidity.

\begin{equation*}
f(x,y) = \sin(x)*y^2 + \cos(y)
\end{equation*}


## Vector Field

A vector field is a function that assigns a vector (i.e., a quantity that has both magnitude and direction) to each point in space. In physics, a vector field is used to describe how a given vector quantity changes in space from point to point, for example, the velocity of a fluid, or an electric or magnetic field. An example of defining a vector function in 2D (the components of the vectors are defined using scalar functions)

\begin{equation*}
    \mathbf{F}(x,y) =  F_x(x,y) \, \mathbf{i} + F_y(x,y) \, \mathbf{j} =  ( x^2-y ) \, \mathbf{i}   + ( \sin(x) + y^2 )  \, \mathbf{j} 
\end{equation*}

## The  $\nabla$ Operator


The **nabla operator** is a vector operator introduced in 1837 by the Irish mathematician and physicist [William Rowan Hamilton](https://en.wikipedia.org/wiki/Nabla_symbol). In Cartesian coordinates, the operator is formally defined as a vector:

\begin{equation*}
\nabla \equiv  \frac{\partial}{\partial x} \, \mathbf{i} + \frac{\partial}{\partial y} \, \mathbf{j} +  \frac{\partial}{\partial z} \, \mathbf{k}
\end{equation*}

By a coordinate transformation, it is possible to obtain the $\nabla$ operator in other coordinate systems (cylindrical, spherical).

## Gradient

By applying the $\nabla$ operator to a scalar field $f(x,y)$
 we obtain a vector field that describes the rate of increase or decrease of the scalar function. 

\begin{equation*}
\operatorname{grad} \, f(x,y,z) = \nabla f(x,y, z) =  \frac{\partial f}{\partial x} \, \mathbf{i} + \frac{\partial f}{\partial y} \, \mathbf{j} + \frac{\partial f}{\partial z} \, \mathbf{k}
\end{equation*}

## Divergencia

Skalárnym súčin operátora $\nabla$ s vektorovým poľom dostaneme skalárne pole, ktoré popisuje žriedlovosť vektorového poľa. Ak v danom bode priestoru pole vzniká, divergencia má kladnú hodnotu, ak zaniká má hodnotu zápornú.  

\begin{equation*}
div \, \mathbf{F}(x,y,z) = \nabla \cdot  \mathbf{F}(x,y,z) = \frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y} + \frac{\partial F_z}{\partial z}
\end{equation*}

## Curl

By taking the cross product of the $\nabla$ operator with a vector field, we obtain a vector field that describes the amount of rotation (or “vorticity”) of the vector field, i.e., the extent to which the field rotates at a given point in space. The curl vector at a given point points along the axis of rotation, and its orientation relative to the plane of rotation is determined using the right-hand rule.

\begin{equation*}
curl \, \mathbf{F}(x,y,z) = \nabla \times \mathbf{F}(x,y,z) 
= \left(\frac{\partial F_z}{\partial y} - \frac{\partial F_y}{\partial z}\right) \mathbf{i}   +  \left(\frac{\partial F_x}{\partial z} - \frac{\partial F_z}{\partial x} \right) \mathbf{j} + \left(\frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y} \right) \mathbf{k} 
\end{equation*}

## Gauss’s Theorem

Gauss’s theorem (the divergence theorem) relates the flux of a vector field $\mathbf{F}$ through a closed smooth surface 
𝑆
forming the boundary of a volume 
𝑉
to the integral of the divergence of the field over that volume:

\begin{equation*}
\oint\limits_S \mathbf{F} \, \mathrm{d}\mathbf{S} = \int\limits_V \nabla \cdot  \mathbf{F} \, \mathrm{d}V 
\end{equation*}



If there is a source  generating the vector field 
$\mathbf{F}$ within the volume 
 $V$  bounded by the surface 
$S$, then the flux integral through the surface is positive; if there is no source, it is zero; and in the case of a sink, it is negative.

## Stokes’ Theorem

Stokes’ theorem relates the line integral around a closed curve $C$ in a vector field  $\mathbf{F}$ to the surface integral of the curl of the field over a surface bounded by that curve.

\begin{equation*}
\oint\limits_C \mathbf{F}\cdot\mathrm{d}\mathbf{l} = \int\limits_S \left({\nabla \times \mathbf{F}}\right) \cdot \mathrm{d}\mathbf{s} 
\end{equation*}

Stokes’ theorem in two-dimensional space is called **Green’s theorem**.
