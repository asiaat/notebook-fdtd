## Transformations of Elementary Quantities

### Length

All units used in *meep* are relative. The basis for converting quantities into real-world values is the unit of length, **`a`**, in which we specify the dimensions of the simulation area, the sizes of geometric objects, and so on.

$$
[l] = a
$$

For example, one unit of length in *meep* (`a`) can be chosen to represent any physical distance, such as $a = 1 \, \mu\text{m}$ or $a = 5 \, \text{cm}$.

### Time

In *meep*, the speed of light in a vacuum is set to have the value **`c = 1`**. Consequently, an electromagnetic wave travels one unit of length (`a`) in exactly one unit of time.

The MEEP unit of time, `[t]`, is therefore related to the unit of length `a` by:

$$
[t] = \frac{a}{c}
$$

Since `c` is set to 1 in MEEP units, the time unit is numerically equal to the length unit.

Furthermore, because the speed of light is defined by the vacuum permittivity ($\epsilon_0$) and permeability ($\mu_0$):

$$
c = \frac{1}{\sqrt{\epsilon_0 \mu_0}}
$$

Setting `c = 1` implies that in MEEP's unit system, $\epsilon_0 = 1$ and $\mu_0 = 1$.



### Current and Mass

To transform derived quantities, it is necessary to know the transformation relationships for current and mass.

In *meep*, the unit of current is set to **`I = 1`**. We can derive the transformation relationship for the unit of mass by starting with Coulomb's Law.

First, we express Coulomb's Law in terms of its fundamental physical units (ignoring the dimensionless $4\pi$ factor):

$$
F = \frac{1} {4 \pi \epsilon_0} \frac{Q_1 Q_2} {r^2} \quad \longrightarrow \quad  [ \text{kg} \, \text{m} \, \text{s}^{-2} ] = \frac{1}{\epsilon_0} \left[ \frac{(\text{A} \cdot \text{s})^2}{\text{m}^2} \right]
$$

Next, we substitute the MEEP base units for Mass (`M`), Length (`a`), Time (`t`), and Current (`I`):

$$
M \cdot a \cdot t^{-2} = \frac{I^2 \cdot t^2}{\epsilon_0 \cdot a^2}
$$

Finally, by solving for `M`, we find the MEEP unit of mass. The expression can be simplified further by using the relationship $t = a/c$:

$$
M = \frac{I^2 \cdot t^4}{\epsilon_0 \cdot a^3} = \frac{I^2 \cdot a}{\epsilon_0 \cdot c^4}
$$

Since `I`, `c`, and `ε₀` are all set to 1 in MEEP's unit system, the unit of mass is numerically equal to the unit of length `a`.



## Transformations of Derived Quantities

### Electric Field Intensity (E)

It follows from the definition of the electric field ($E = F/Q$). The MEEP unit for `E` is derived by substituting the MEEP units for force and charge:

$$
[E] = \frac{[F]}{[Q]} \quad \longrightarrow \quad \frac {M \cdot a } {I \cdot t^3} = \frac{I^2 \cdot a}{\epsilon_0 \cdot c^4} \frac {a } {I \cdot t^3} = \frac {I } {\epsilon_0 \cdot a \cdot c}
$$

The transformation relationship for converting a unitless `E` value from a MEEP simulation to the SI system is:

$$
E_{SI} = E_{\text{MEEP}} \cdot \frac{I_{SI}} {a_{SI}} \cdot (3.7673 \times 10^{2}) \, \left[ \frac{\text{V}}{\text{m}} \right]
$$

*Note: The constant is the approximate value of the impedance of free space, $Z_0$.*

### Electric Displacement (D)

From the relationship $D = \epsilon_0 E$, the MEEP unit for `D` is derived:

$$
[D] = [\epsilon_0] [E] \quad \longrightarrow \quad \frac {I } { a \cdot c}
$$

The transformation relationship for converting a unitless `D` value from a MEEP simulation to the SI system is:

$$
D_{SI} = D_{\text{MEEP}} \cdot \frac{I_{SI}} {a_{SI}} \cdot (3.3356 \times 10^{-9}) \, \left[ \frac{\text{C}}{\text{m}^2} \right]
$$
