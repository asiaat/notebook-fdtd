# Scalar Field


In mathematics and physics, a scalar field is a function $f(x,y,z)$
that assigns a scalar value (i.e., a value specified by a single numerical quantity) to every point in space. $(x,y,z)$. 

The values of a scalar field can be real or complex. Examples of scalar fields include, for instance, temperature, pressure, density, or humidity in meteorology, and electric potential in electrical engineering.

## Numerical Calculation

To create a scalar field, we need to generate a 2D or 3D array of values (the domain) in numpy, over which the scalar field will be defined.

    dx = np.linspace(-1.0, 1.0, 30)  # range and number of samples along the x-axis
    dy = np.linspace(-1.0, 1.0, 30)  # and along the y-axis
    x,y = np.meshgrid(dx, dy)        # 2D grid of values

The function \textit{meshgrid()} generates the arrays 
𝑥
,
𝑦
x,y; these contain values that represent a grid of points $(x_0, y_0), (x_0, y_1) \dots$.  In the following example, let’s create and visualize a 2D scalar field:

\begin{equation*}
f(x,y) = x^2 + \frac{y^3}{2}
\end{equation*}

<!-- MEDSKIP -->

To visualize a scalar field, we can use the *contour()* and *plot_surface()* methods from the matplotlib library. The values of a 2D scalar function are displayed using a color scale and contour lines.
<!-- MEDSKIP -->

```python
%reset -sf 
%matplotlib inline

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

dx = np.linspace(-1.0, 1.0, 30)      # coordinate ranges
dy = np.linspace(-1.0, 1.0, 30) 
x,y = np.meshgrid(dx, dy)            # coordinate arrays (xi, yi) ...

def f(x,y):                          # definition of a 2D scalar function
    return (x**2 + y**3/2) 

fig = plt.figure(figsize=(7, 5))     # display of the scalar field
ax = fig.add_subplot(111)
img = ax.imshow(f(x,y), extent=[-1, 1, -1, 1], 
                origin='lower', interpolation='kaiser', 
                clip_on=True, alpha=0.5, cmap='jet') 
fig.colorbar(img)

ax.contour(x, y, f(x,y), 20, colors="k", alpha=0.2) #, offset=-0.50)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Skalar Field')
plt.grid()
plt.show() 
```
