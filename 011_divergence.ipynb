


# Divergence

Divergence is a scalar quantity that describes the **source-ness** or **sink-ness** of a vector field; in other words, whether the field flows **into** a point (sink) or **out of** a point (source).

The divergence of a vector field $\mathbf{F}$ is defined as:

\begin{equation}
\operatorname{div} \, \mathbf{F} = \nabla \cdot \mathbf{F} = 
\frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y} + \frac{\partial F_z}{\partial z}
\end{equation}

**Explanation of variables:**

- $\mathbf{F} = (F_x, F_y, F_z)$ is the vector field, with components along the $x$, $y$, and $z$ axes.  
- $F_x, F_y, F_z$ are the scalar functions representing the vector field components in the respective directions.  
- $\nabla \cdot \mathbf{F}$ (divergence) gives a **scalar field** that quantifies how much the vector field is "spreading out" (positive divergence) or "converging" (negative divergence) at each point.  

**Additional Note:**  

- Divergence is widely used in physics, for example in **fluid dynamics** to measure fluid sources/sinks, and in **electromagnetism** in Gauss's law to relate electric flux to charge density.




## Divergence in Cylindrical Coordinates

The divergence of a vector field $\mathbf{F}$ in cylindrical coordinates $(r, \theta, z)$ is given by:

\begin{equation}
\operatorname{div} \mathbf{F} = \nabla \cdot \mathbf{F} = 
\frac{1}{r} \frac{\partial}{\partial r} \left(r F_r\right) 
+ \frac{1}{r} \frac{\partial F_\theta}{\partial \theta} 
+ \frac{\partial F_z}{\partial z}
\end{equation}

**Explanation of variables:**

- $\mathbf{F} = (F_r, F_\theta, F_z)$ is the vector field in **cylindrical coordinates**, with components along the radial ($r$), angular ($\theta$), and vertical ($z$) directions.  
- $F_r$ is the radial component, $F_\theta$ is the angular component, and $F_z$ is the vertical component.  
- The factor $1/r$ appears in the radial and angular terms to account for the geometry of cylindrical coordinates (radial distance and circular arc length).  
- $\nabla \cdot \mathbf{F}$ gives a **scalar field** representing the net "outflow" or "inflow" of the vector field at each point.

**Additional Note:**  

- Positive divergence indicates a **source** (field spreading out), negative divergence indicates a **sink** (field converging), and zero divergence indicates an **incompressible field** in that region.







## Properties of Divergence

If $\mathbf{F}, \mathbf{G}$ are vector fields, $\varphi$ is a scalar field, and $a, b$ are real numbers, then the divergence operator satisfies the following identities:

- **Linearity:**

\begin{equation*}
\operatorname{div}( a\mathbf{F} + b\mathbf{G} ) = a \operatorname{div} \mathbf{F} + b \operatorname{div} \mathbf{G}
\end{equation*}

This means the divergence of a linear combination of vector fields equals the same linear combination of their divergences.

- **Divergence of a scalar times a vector field:**

\begin{equation*}
\operatorname{div}(\varphi \mathbf{F}) = \operatorname{grad} \varphi \cdot \mathbf{F} + \varphi \operatorname{div} \mathbf{F}
\end{equation*}

or equivalently:

\begin{equation*}
\nabla\cdot(\varphi \mathbf{F}) = (\nabla\varphi) \cdot \mathbf{F} + \varphi (\nabla\cdot\mathbf{F})
\end{equation*}

This is the **product rule** for divergence.

- **Divergence of the cross product of two vector fields:**

\begin{equation*}
\operatorname{div}(\mathbf{F}\times\mathbf{G}) = \operatorname{curl} \mathbf{F} \cdot\mathbf{G} - \mathbf{F} \cdot \operatorname{curl} \mathbf{G}
\end{equation*}

or equivalently:

\begin{equation*}
\nabla\cdot(\mathbf{F}\times\mathbf{G}) = (\nabla\times\mathbf{F})\cdot\mathbf{G} - \mathbf{F}\cdot(\nabla\times\mathbf{G})
\end{equation*}

- **Divergence of the gradient (Laplacian):**

\begin{equation*}
\operatorname{div}(\nabla\varphi) = \Delta\varphi
\end{equation*}

Here, $\Delta \varphi$ is the **Laplacian** of the scalar field $\varphi$.

- **Divergence of the curl is zero:**

\begin{equation*}
\nabla\cdot(\nabla\times\mathbf{F}) = 0
\end{equation*}

**Explanation of variables:**

- $\mathbf{F}, \mathbf{G}$: vector fields  
- $\varphi$: scalar field  
- $\nabla\cdot$: divergence operator  
- $\nabla\times$: curl operator  
- $\nabla\varphi$: gradient of a scalar field  
- $\Delta\varphi$: Laplacian of a scalar field  

These identities are fundamental in **vector calculus**, physics (electromagnetism, fluid dynamics), and engineering.




## Discrete Approximation of Divergence

The discrete approximation of divergence in the 2D plane $(x,y)$ can be derived from the components of a vector field on the edges of a small cell of size $dx \times dy$.  
We assume that the values of the vector field components are constant along each edge of the cell.

<img src="./img/div_2d_01.png" width=400px alt="Divergence in the x-y plane" scale="1.25">

For the divergence, we can then approximately write (as $\Delta x, \Delta y \rightarrow 0$):

\begin{equation*}
\frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y} \approx
\frac{F_x\!\left(x +\tfrac{\Delta x}{2}, y\right) - F_x\!\left(x -\tfrac{\Delta x}{2}, y\right)}{\Delta x} +
\frac{F_y\!\left(x, y +\tfrac{\Delta y}{2}\right) - F_y\!\left(x, y -\tfrac{\Delta y}{2}\right)}{\Delta y} 
\end{equation*}

If we take $\Delta x = \Delta y = \delta$, the expression simplifies to:

\begin{equation}
\frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y} \approx
\frac{1}{\delta} \Big(
F_x\!\left(x +\tfrac{\delta}{2}, y\right) - F_x\!\left(x -\tfrac{\delta}{2}, y\right) + 
F_y\!\left(x, y +\tfrac{\delta}{2}\right) - F_y\!\left(x, y -\tfrac{\delta}{2}\right) \Big)
\end{equation}

A similar approach can be used to derive the discrete approximation of divergence in **3D space**.



### Explanation of Variables:
- $F_x, F_y$: components of the vector field $\mathbf{F}(x,y)$ in the $x$ and $y$ directions  
- $\Delta x, \Delta y$: cell dimensions in the $x$ and $y$ directions  
- $\delta$: uniform grid spacing when $\Delta x = \Delta y$  

This formulation is the basis of **finite-difference methods (FDM)** used in numerical simulations of fluid flow, electromagnetism, and other fields of computational physics.


## Symbolic Computation

The symbolic mathematics library *sympy* provides the function `divergence()` that can be used to compute the divergence of a vector field.  
For example, in the 2D plane, consider the vector field:

\begin{equation*}
\mathbf{F}(x,y) =  
\mathbf{i}\,\Big( -y (1-x-y)+ \tfrac{x}{4} \Big)  
+ \mathbf{j}\,\Big( 2 -x^2 + \tfrac{y}{4} \Big)
\end{equation*}

where:  
- $\mathbf{i}$ is the unit vector in the $x$-direction,  
- $\mathbf{j}$ is the unit vector in the $y$-direction,  
- the expressions inside the parentheses are the scalar components $F_x(x,y)$ and $F_y(x,y)$ of the vector field $\mathbf{F}$.  

The divergence is then given by:

\begin{equation*}
\nabla \cdot \mathbf{F}(x,y) = 
\frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y}
\end{equation*}

Using `sympy.divergence()`, we can compute this automatically and simplify the result.


```python
from IPython.display import *
from IPython.display import Latex as lx
from sympy import latex

def ltxprint(var, value):   
    # pomocna funkcia pre formatovanu tlac premennych
    s = '$' + latex(var) + '=' + latex(value) + '$'
    #display(lx(s))
    display(s)
```

```python



import os
os.getcwd()


from sympy import *
from sympy.vector import CoordSys3D, divergence

r = CoordSys3D('r')                  # suradnicova sustava
F = (-r.y*(1 - r.x - r.y) + r.x/4)*r.i + (2 - r.x**2 + r.y/4)*r.j
D = divergence(F)

ltxprint(r'\nabla \cdot \mathbf{F}', D)
```

Value of the Divergence at a Chosen Point $P(x,y) = (-0.5, 0.5)$

```python
P = [-0.5, 0.5]
value = D.subs([ (r.x, P[0]), (r.y, P[1]) ])
ltxprint(r'\mathrm{div}(P)', value)
```



With divergence, we can work as with a standard scalar field.  

In our case, the result of the symbolic computation of divergence in *sympy* is visualized using the libraries *matplotlib* and *numpy* through formal substitution of variables and evaluation of expressions.


```python
from sympy import *
from sympy.vector import CoordSys3D, divergence

# Formal substitution of variables, (x, y) – are not variables in *sympy*, but strings.

Fx = F.components[r.i].subs([(r.x, 'x'), (r.y, 'y')])                 
Fy = F.components[r.j].subs([(r.x, 'x'), (r.y, 'y')]) 
Ds = D.subs([(r.x, 'x'), (r.y, 'y')])   

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

dx = np.linspace(-1.0, 1.0, 20)    # Coordinate ranges
dy = np.linspace(-1.0, 1.0, 20)
x, y = np.meshgrid(dx, dy) 

a = eval(str(Fx))                  # Numerical computation of components
b = eval(str(Fy))                  # Evaluation of strings – conversion from *sympy*
d = eval(str(Ds))

fig = plt.figure(figsize=(7, 5)) 
ax = fig.add_subplot(111)
ax.streamplot(x, y, a, b, color='r')
img = ax.imshow(d, extent=[-1, 1, -1, 1], origin='lower', interpolation='kaiser', 
                 clip_on=True, alpha=0.5, cmap='jet') 
fig.colorbar(img)
plt.grid() 
plt.show()
```




## Numerical Calculation

Using the *numdifftools* library, we can compute the divergence of the vector field $\mathbf{F}(x,y)$ from the previous example at the point in the plane $P(x,y) = (-0.5, 0.5)$.  

For the 2D case, the Jacobian matrix has the form:

\begin{equation*}
\mathbf{J} 
= \begin{bmatrix}
    \dfrac{\partial F_x}{\partial x} &  \dfrac{\partial F_x}{\partial y} \\
    \dfrac{\partial F_y}{\partial x} &  \dfrac{\partial F_y}{\partial y} 
  \end{bmatrix}
\end{equation*}

**Explanation of variables:**

- $\mathbf{F}(x,y) = (F_x(x,y), F_y(x,y))$: the 2D vector field, with components in the $x$ and $y$ directions.  
- $F_x, F_y$: scalar component functions of the vector field.  
- $P(x,y) = (-0.5, 0.5)$: the point at which the divergence is evaluated.  
- $\mathbf{J}$: the **Jacobian matrix** of $\mathbf{F}$, whose diagonal elements are the partial derivatives of the components with respect to their own variable, and off-diagonal elements are cross derivatives.  
- The divergence in 2D can then be obtained as the **trace** of the Jacobian:  
  \[
  \nabla \cdot \mathbf{F} = \frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y}
  \]

```python
#%reset -f 
import numdifftools as nd
import numpy as np

def F(p):
    [x,y] = p
    return np.array([ (-y*(1 - x - y**2) + x/4),  (2 - x**2 + y/4) ])

P = [-0.5, 0.5]
J = nd.Jacobian(F)
value = J(P)
print(value)
```



The value of the divergence can be determined directly using the elements of the Jacobian matrix:

\begin{equation*}
\mathrm{div} \, \mathbf{F} = \frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y} 
\end{equation*}

at the point in the plane $P(x,y) = (-0.5, 0.5)$.


```python
div = value[0,0] + value[1,1]
print(div)
```
