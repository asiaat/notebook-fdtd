# Vector Field

## Properties

In mathematics and physics, a vector field is a function that assigns a vector (i.e., a quantity with both magnitude and direction) to each point in space. In physics, a vector field is used to describe how a given vector quantity changes from point to point in space—for example, the velocity of a fluid in hydrodynamics, the electric or magnetic field in electronics, or the direction and speed of the wind in meteorology.

In 3D space, a vector field is defined as:

\begin{equation}
\mathbf{F}(x,y,z) =  F_x(x,y,z) \, \mathbf{i}  +  F_y(x,y,z) \, \mathbf{j}  +  F_z(x,y,z) \, \mathbf{k}  
\end{equation}


## Numerical Calculation

 We can define vector Field $\mathbf{F}(x,y)$ in 2D,  using scalar functions $F_x(x,y)$ and $F_y(x,y)$ for the individual components of the field.

\begin{equation*}
\mathbf{F}(x,y) =  F_x(x,y) \, \mathbf{i} +  F_y(x,y) \, \mathbf{j} =  \left( -y \, (1-x-y)+ \frac{x}{4} \right) \, \mathbf{i}  +  \left(2 -x^2 + \frac{y}{4} \right) \, \mathbf{j} 
\end{equation*}

<!-- MEDSKIP -->
To visualize a vector field, we can use the *quiver()* method from the *matplotlib* library.
<!-- MEDSKIP -->

```python
%reset -sf 
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

dx = np.linspace(-1.0, 1.0, 10)             # rozsahy suradnic
dy = np.linspace(-1.0, 1.0, 10)
x, y = np.meshgrid(dx, dy)                  # definicny obor vektoroveho pola

[Fx, Fy] = [ -y*(1-x-y)+ x/4, 2-x**2+y/4 ]  # zlozky vektoroveho pola

fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(111)
ax.quiver(dx, dy, Fx, Fy, color='r')
plt.xlabel('x')
plt.ylabel('y')
plt.axis('equal')
plt.title('Vektorove pole')
plt.grid()
plt.show()
```
