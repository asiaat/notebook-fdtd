# Maxwell's Equations

## Static Fields

In the static case (where physical quantities do not change over time), the electric and magnetic fields are independent. In simple terms, a stationary charge does not affect a steady current flowing through a conductor, and conversely, a steady current does not exert a force on a static charge. Maxwell's equations in this scenario describe the character of the field, i.e., whether it is **source-like (irrotational)** or **vortex-like (solenoidal)**.

$$
\begin{align*}
\nabla \cdot \mathbf{D} &= \rho_v \\
\nabla \cdot \mathbf{B} &= 0 \\
\nabla \times \mathbf{E} &= 0 \\
\nabla \times \mathbf{H} &= \mathbf{J} \\
\end{align*}
$$

### Explanation of the Equations

1.  **Gauss's Law for Electricity**: $\nabla \cdot \mathbf{D} = \rho_v$
    * $\mathbf{D}$ is the electric displacement field.
    * $\rho_v$ is the volume charge density.
    * This equation states that electric charges are the sources (or sinks) of the electric field.

2.  **Gauss's Law for Magnetism**: $\nabla \cdot \mathbf{B} = 0$
    * $\mathbf{B}$ is the magnetic field (or magnetic induction).
    * The zero divergence indicates that there are no magnetic monopoles; magnetic field lines always form closed loops.

3.  **Faraday’s Law (static case)**: $\nabla \times \mathbf{E} = 0$
    * In the static case, the electric field is conservative (there is no changing magnetic field).
    * This means the electric field can be expressed as the gradient of a potential: $\mathbf{E} = - \nabla V$.

4.  **Ampère’s Law (static case)**: $\nabla \times \mathbf{H} = \mathbf{J}$
    * $\mathbf{H}$ is the magnetic field intensity.
    * $\mathbf{J}$ is the current density vector.
    * A steady current creates a magnetic field with circulation (a "curl") around the current.

### Lorentz Force

A charge $Q$ moving with a constant velocity $\mathbf{v}$ in an electric and magnetic field experiences a force:

$$
\mathbf{F} = Q (\mathbf{E} + \mathbf{v} \times \mathbf{B})
$$

* $\mathbf{E}$ — electric field
* $\mathbf{B}$ — magnetic field
* $\mathbf{v} \times \mathbf{B}$ — magnetic force (perpendicular to both velocity and field)

**Example 1: Point Charge in a Static Electric Field**
* If $\mathbf{B} = 0$ and $\mathbf{E} \neq 0$, the force is simply $\mathbf{F} = Q \mathbf{E}$.
* A positive charge accelerates along the field lines; a negative charge accelerates in the opposite direction.

**Example 2: Charged Particle in a Magnetic Field**
* If $\mathbf{E} = 0$ and $\mathbf{B} \neq 0$, the particle moves in a circular path perpendicular to $\mathbf{B}$ due to the Lorentz force.
* The radius of the circle is given by $r = \frac{m v}{Q B}$, where $m$ is the particle's mass and $v$ is its speed.

These static Maxwell equations are the foundation of **electrostatics** and **magnetostatics**.

# Time-Varying Fields

When charges are accelerating or when currents change with time, the electric and magnetic fields are no longer independent. They become dynamically coupled: a changing magnetic field creates an electric field, and a changing electric field creates a magnetic field. This interplay is the foundation of electromagnetic waves (like light, radio waves, and X-rays) and is described by the full set of Maxwell's Equations.

---

### 1. Faraday's Law of Induction: How B Creates E

A changing magnetic flux through a surface induces a circulating electric field around the boundary of that surface. This is the principle behind electric generators and transformers.

**Differential Form:** $\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}$

* **Concept**: This equation states that a time-varying magnetic field ($\frac{\partial \mathbf{B}}{\partial t}$) is the source of a "swirling" or circulating electric field ($\nabla \times \mathbf{E}$). Unlike the static E-field, this induced E-field is **non-conservative**—the work done moving a charge in a closed loop is not zero.
* **Symbols**:
    * $\nabla \times \mathbf{E}$: The curl of the electric field.
    * $-\frac{\partial \mathbf{B}}{\partial t}$: The negative rate of change of the magnetic field with respect to time. The negative sign (Lenz's Law) indicates that the induced E-field creates a current that opposes the change in magnetic flux.

**Integral Form:** $\oint_L \mathbf{E} \cdot d\mathbf{l} = -\frac{d}{dt} \int_S \mathbf{B} \cdot d\mathbf{s}$

* **Concept**: The electromotive force (voltage) induced around a closed loop $L$ is equal to the negative rate of change of the magnetic flux passing through the surface $S$ bounded by that loop.
* **Example**: If you move a magnet towards a coil of wire, the changing magnetic flux through the coil induces a voltage ($\oint_L \mathbf{E} \cdot d\mathbf{l}$), causing a current to flow.

---

### 2. Ampère's Law with Maxwell's Correction: How E and J Create B

A magnetic field is created by two sources: moving charges (conduction current, **J**) and a time-varying electric field. The second source, added by Maxwell, is called the **displacement current**.

**Differential Form:** $\nabla \times \mathbf{H} = \mathbf{J} + \frac{\partial \mathbf{D}}{\partial t}$

* **Concept**: This law states that the circulation of the H-field ($\nabla \times \mathbf{H}$) is sourced by both the flow of charge ($\mathbf{J}$) and the rate of change of the electric displacement field ($\frac{\partial \mathbf{D}}{\partial t}$).
* **The Displacement Current ($\frac{\partial \mathbf{D}}{\partial t}$)**: This term was Maxwell's crucial addition. It ensures that the law is consistent with the conservation of charge. It means that even in a vacuum where there is no current ($\mathbf{J}=0$), a changing electric field can still generate a magnetic field.
* **Example**: Consider a capacitor being charged. A current flows in the wires leading to the plates, but no charge flows across the gap between them. However, as charge builds up, the electric field in the gap changes with time. This changing E-field acts as a "displacement current," creating a magnetic field that circles the gap, just as if a real current were flowing.

---

### The Complete Set of Maxwell's Equations (Time-Varying)

These four equations, along with the Lorentz force law, form the complete foundation of classical electromagnetism.

$$
\begin{align*}
\nabla \cdot \mathbf{D} &= \rho_v && \text{(Gauss's Law for E - unchanged)} \\
\nabla \cdot \mathbf{B} &= 0 && \text{(Gauss's Law for B - unchanged)} \\
\nabla \times \mathbf{E} &= -\frac{\partial \mathbf{B}}{\partial t} && \text{(Faraday's Law)} \\
\nabla \times \mathbf{H} &= \mathbf{J} + \frac{\partial \mathbf{D}}{\partial t} && \text{(Ampère-Maxwell Law)} \\
\end{align*}
$$

### Consequence: The Wave Equation

The coupling of the E and B fields in the curl equations leads directly to the prediction of electromagnetic waves. In a source-free region ($\rho_v=0, \mathbf{J}=0$), the equations can be combined to form the wave equation for the electric field:

$$
\nabla^2 \mathbf{E} - \mu\epsilon \frac{\partial^2 \mathbf{E}}{\partial t^2} = 0
$$

* **Concept**: This is a standard wave equation that describes a disturbance propagating through space.
* **Speed of Propagation**: The equation predicts that this disturbance—an electromagnetic wave—travels at a speed of $c = \frac{1}{\sqrt{\mu\epsilon}}$. In a vacuum, this speed is the speed of light.
