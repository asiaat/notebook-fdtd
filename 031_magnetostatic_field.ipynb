

# Magnetostatic Field

## Magnetic Induction

A magnetic field forms closed loops, which do not start or end at any point in space, because magnetic charges (monopoles) do not exist. The magnetic field is described by the magnetic induction $\mathbf{B}$. A moving charge in a magnetic field experiences a force (Lorentz force):

\begin{equation*}
\mathbf{F} = Q \, (\mathbf{v} \times \mathbf{B})
\end{equation*}

The source of a magnetic field is moving charges—electric current. A current $I$ flowing through a segment of a conductor of length $\mathrm{d} \mathbf{l}$ generates a magnetic field in its vicinity (Biot-Savart law):

\begin{equation*}
\mathrm{d} \mathbf{B} = \frac{\mu_0}{4 \pi} \frac{I \, \mathrm{d} \mathbf{l} \times \mathbf{r}}{r^3}
\end{equation*}

where $\mathbf{r}$ is the position vector from the conductor segment to the observation point. Because, due to charge conservation, electric current must flow in a closed circuit, the magnetic field of a current loop is:

\begin{equation*}
\mathbf{B} = \frac{\mu_0}{4 \pi} \oint\limits_L \frac{I \, \mathrm{d} \mathbf{l} \times \mathbf{r}}{r^3}
\end{equation*}

### Variables

- $\mathbf{B}$ — magnetic induction (magnetic field)  
- $\mathbf{F}$ — Lorentz force on a charge  
- $Q$ — moving electric charge  
- $\mathbf{v}$ — velocity of the charge  
- $I$ — electric current  
- $\mathrm{d} \mathbf{l}$ — differential element of the current-carrying conductor  
- $\mathbf{r}$ — vector from the conductor element to the observation point  
- $r$ — magnitude of $\mathbf{r}$  
- $\mu_0$ — permeability of free space  
- $L$ — path of the current loop




## Ampère's Law

The relationship between electric current and magnetic induction is expressed by Ampère's law:

\begin{equation*}
\oint\limits_L \mathbf{B} \cdot \mathrm{d} \mathbf{l} = \mu_0 I_c
\end{equation*}

where $I_c$ is the total current passing through the surface enclosed by the curve $L$.

Using Stokes' theorem, we can rewrite the line integral as a surface integral:

\begin{equation*}
\oint\limits_L \mathbf{B} \cdot \mathrm{d} \mathbf{l} = \int\limits_S \nabla \times \mathbf{B} \, \mathrm{d} \mathbf{s}
= \mu_0 I_c = \mu_0 \int\limits_S \mathbf{J} \, \mathrm{d} \mathbf{s}
\end{equation*}

Comparing the integrands gives Ampère's law in differential form:

\begin{equation*}
\nabla \times \mathbf{B} = \mu_0 \, \mathbf{J} 
\end{equation*}

### Variables

- $\mathbf{B}$ — magnetic induction (magnetic field)  
- $\mathrm{d} \mathbf{l}$ — differential element along the curve $L$  
- $\mu_0$ — permeability of free space  
- $I_c$ — total current passing through surface $S$  
- $\mathbf{J}$ — current density vector  
- $S$ — surface bounded by curve $L$  
- $L$ — closed curve along which the magnetic field is integrated
