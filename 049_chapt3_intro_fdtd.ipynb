## Introduction to the FDTD Method

The **Finite-Difference Time-Domain (FDTD)** method is a numerical technique for solving Maxwell's equations. It is conceptually simple and can accurately model a wide range of complex electromagnetic problems, though it can be computationally expensive. It is best suited for "resonance region" problems, where the size of the objects being modeled is roughly on the same order of magnitude as the wavelength of the electromagnetic waves.

The core principle of FDTD is to replace the continuous spatial and temporal derivatives in Maxwell's equations (specifically Ampere's and Faraday's laws) with discrete **finite-difference approximations**.



### The Central-Difference Approximation

The method derives a highly accurate approximation for a derivative using Taylor series expansions. By expanding a function $f(x)$ around a point $x_0$ with a small offset $\pm\delta/2$:

$$
f(x_{0}+\frac{\delta}{2})=f(x_{0})+\frac{\delta}{2}f^{\prime}(x_{0})+\frac{1}{2!}\left(\frac{\delta}{2}\right)^{2}f^{\prime\prime}(x_{0})+...
$$
$$
f(x_{0}-\frac{\delta}{2})=f(x_{0})-\frac{\delta}{2}f^{\prime}(x_{0})+\frac{1}{2!}\left(\frac{\delta}{2}\right)^{2}f^{\prime\prime}(x_{0})-...
$$

Subtracting these two equations and solving for the derivative $f'(x_0)$ results in the **central-difference approximation**:

$$
\frac{df(x)}{dx}\bigg|_{x=x_{0}} = \frac{f(x_{0}+\frac{\delta}{2})-f(x_{0}-\frac{\delta}{2})}{\delta} + O(\delta^{2})
$$

**Explanation of the Equation:**

* **$\frac{df(x)}{dx}\big|_{x=x_{0}}$**: The exact derivative of the function $f(x)$ at the point $x_0$.
* **$\frac{f(x_{0}+\frac{\delta}{2})-f(x_{0}-\frac{\delta}{2})}{\delta}$**: The finite-difference approximation. It calculates the derivative at a central point ($x_0$) using the function's values at two neighboring points, each a half-step ($\delta/2$) away.
* **$O(\delta^{2})$**: This term represents the error of the approximation. Because the leading error term is proportional to the square of the step size ($\delta^2$), this method is said to have **second-order accuracy**. This means that if you halve the step size, the error is reduced by a factor of four, making it very efficient.
