# Gauss's Theorem

Gauss's theorem (also called the Gauss-Ostrogradsky theorem or the divergence theorem) relates the **flux** of a vector field $\mathbf{F}$ through a closed surface $S$, which bounds a volume $V$, to the **sources** of the field contained inside the volume.

\begin{equation}
\int\limits_V \nabla \cdot \mathbf{F} \, dV =  \oint\limits_S \mathbf{F} \, d\mathbf{S}
\end{equation}

where the vector field is defined as:

\begin{equation*}
\mathbf{F}(x,y,z) = F_x(x,y,z) \, \mathbf{i}  +  F_y(x,y,z) \, \mathbf{j} +  F_z(x,y,z) \, \mathbf{k}
\end{equation*}

---

**Explanation of variables:**

- $\mathbf{F} = (F_x, F_y, F_z)$: the 3D vector field  
- $F_x, F_y, F_z$: scalar component functions of $\mathbf{F}$ in the $x$, $y$, and $z$ directions  
- $V$: the volume over which the divergence is integrated  
- $S$: the closed surface bounding the volume $V$  
- $dV$: differential volume element  
- $d\mathbf{S}$: oriented surface element vector  
- $\nabla \cdot \mathbf{F}$: divergence of $\mathbf{F}$, representing the net source strength at each point in the volume





## Discrete Approximation of Gauss's Theorem

We divide the volume $V$ into small cubes with edge lengths $\mathrm{d} x \, \mathrm{d} y \, \mathrm{d} z$. The oriented surface element $\mathrm{d} \mathbf{S}$ is perpendicular to the surface and has magnitude equal to the area of that surface:

\begin{equation*}
\mathrm{d} \mathbf{S} = \mathbf{n} \, \mathrm{d} S, \quad \mathbf{n} \perp S
\end{equation*}

**TODO: Add illustration**

\begin{align*}
\mathrm{d} \mathbf{S_x} &= \mathbf{i} \, S_x  = \mathbf{i} \, \mathrm{d} y \,  \mathrm{d} z\\
\mathrm{d} \mathbf{S_y} &= \mathbf{j} \, S_y  = \mathbf{j} \, \mathrm{d} x \, \mathrm{d} z \\
\mathrm{d} \mathbf{S_z} &= \mathbf{k} \, S_z  = \mathbf{k} \, \mathrm{d} x \, \mathrm{d} y
\end{align*}

**TODO: Add illustration**

The divergence of the vector field can then be approximated as:

\begin{align*}
\int\limits_V \nabla \cdot \mathbf{F} \, \mathrm{d} V = \
&\sum_{i,j,k} \left( \frac{F_x(x_{i+1},y_j, z_k) - F_x(x_i,y_j,z_k)}{\mathrm{d} x} \right) \mathrm{d} x \, \mathrm{d} y \, \mathrm{d} z   \, + \\
&\sum_{i,j,k} \left( \frac{F_y(x_i,y_{j+1},z_k)  - F_y(x_i,y_j,z_k)}{\mathrm{d} y} \right) \mathrm{d} x \, \mathrm{d} y \, \mathrm{d} z \, + \\
&\sum_{i,j,k} \left( \frac{F_z(x_i,y_j,z_{k+1})  - F_z(x_i,y_j,z_k)}{\mathrm{d} z} \right) \mathrm{d} x \, \mathrm{d} y \, \mathrm{d} z
\end{align*}

If we take the first component of the vector field $F_x$, for fixed indices $j,k$ we can write:

\begin{equation*}
\sum_{i} \left( \frac{F_x(x_{i+1},y_j, z_k) - F_x(x_i,y_j,z_k)}{ \mathrm{d} x} \right) \mathrm{d} x \, \mathrm{d}y \, \mathrm{d} z  
=  \sum_{i} \big( F_x(x_{i+1},y_j, z_k) - F_x(x_i,y_j,z_k)  \big) \, \mathrm{d} y \, \mathrm{d} z
\end{equation*}

---

**Explanation of variables:**

- $\mathbf{F} = (F_x, F_y, F_z)$: the 3D vector field  
- $F_x, F_y, F_z$: components of $\mathbf{F}$ along $x$, $y$, $z$ directions  
- $\mathrm{d} x, \mathrm{d} y, \mathrm{d} z$: grid spacing along $x$, $y$, $z$  
- $\mathrm{d} \mathbf{S}$: oriented surface element vector, perpendicular to the surface  
- $i, j, k$: indices for discretized grid points along $x, y, z$  
- The summations approximate the surface integrals as **finite differences**, representing the flux through each face of the small cube.





If we perform the summation, the values of the field on adjacent cubes will cancel each other, leaving only the flux through the outer faces of the boundary cubes:

\begin{equation*}
\sum_{i} \big( F_x(x_{i+1},y_j, z_k) - F_x(x_i,y_j,z_k)  \big) \, \mathrm{d} y \, \mathrm{d} z 
= \big( F_x(x_{n},y_j, z_k) - F_x(x_1,y_j,z_k)  \big) \, \mathrm{d} y \, \mathrm{d} z
\end{equation*}

---

**Explanation of variables:**

- $F_x(x_i, y_j, z_k)$: $x$-component of the vector field at the discretized point $(x_i, y_j, z_k)$  
- $i$: index along the $x$-axis  
- $x_1, x_n$: coordinates of the first and last cubes along $x$  
- $\mathrm{d}y \, \mathrm{d}z$: area of the surface perpendicular to the $x$-axis  
- The summation shows that **internal fluxes cancel**, leaving only the contribution from the outer surfaces, which is the key idea behind the discrete version of Gauss's theorem.




After simplification, the divergence integral can be rewritten as:

\begin{align*}
\int\limits_V \nabla \cdot \mathbf{F} \, \mathrm{d} V = \
&\sum_{j,k} \big( F_x(x_{n},y_j, z_k) - F_x(x_1,y_j,z_k)  \big) \, \mathrm{d} y \, \mathrm{d} z \, + \\
&\sum_{i,k} \big( F_y(x_i,y_n, z_k) - F_y(x_i,y_1,z_k)  \big) \, \mathrm{d} x \, \mathrm{d} z \, + \\
&\sum_{i,j} \big( F_z(x_i,y_j, z_n) - F_z(x_i,y_j,z_1)  \big) \, \mathrm{d} x \, \mathrm{d} y
\end{align*}

This represents the sum of the fluxes of the vector field through all the outer faces of the cubes forming the volume $V$. From this, it follows:

\begin{equation*}
\int\limits_V \nabla \cdot \mathbf{F} \, \mathrm{d} V = \
\int\limits_{S_x} (F_{x_n} - F_{x_1}) \, \mathrm{d} S_x + \
\int\limits_{S_y} (F_{y_n} - F_{y_1}) \, \mathrm{d} S_y + \
\int\limits_{S_z} (F_{z_n} - F_{z_1}) \, \mathrm{d} S_z 
= \oint\limits_S \mathbf{F} \, \mathrm{d} \mathbf{S}
\end{equation*}

---

**Explanation of variables:**

- $\mathbf{F} = (F_x, F_y, F_z)$: 3D vector field  
- $x_1, x_n, y_1, y_n, z_1, z_n$: coordinates of the boundary faces of the volume along each axis  
- $\mathrm{d}x, \mathrm{d}y, \mathrm{d}z$: grid spacing along $x$, $y$, $z$  
- $S_x, S_y, S_z$: surfaces perpendicular to the $x$, $y$, $z$ axes  
- $\oint\limits_S \mathbf{F} \, \mathrm{d} \mathbf{S}$: total flux of $\mathbf{F}$ through the closed surface $S$  
- This derivation shows that the **sum of fluxes through all outer faces equals the volume integral of divergence**, which is the discrete version of Gauss’s theorem.





## Gauss's Theorem in 2D

For a 2D geometry $(x,y,0)$, Gauss's theorem can be applied using a simple projection onto the plane. In this case, the volume integral reduces to a surface integral, and the surface integral reduces to a line integral along the boundary curve of the surface:

\begin{equation}
\iint\limits_{\Omega} \left( \nabla \cdot \mathbf{F} \right) \, \mathrm{d} x \, \mathrm{d} y 
= \int\limits_{\partial \Omega} \mathbf{F} \, \mathrm{d} \mathbf{l_\perp}
\end{equation}

where:

- $\Omega$ is the 2D surface  
- $\partial \Omega$ is the boundary curve of $\Omega$  
- $\mathrm{d} \mathbf{l_\perp}$ is the vector perpendicular to the line element of the curve  

**TODO: Add illustration**

If the tangent vector of the curve element is $d \mathbf{l} = (dx, dy)$, the components of the perpendicular vector can be determined using the dot product condition for perpendicularity:

\begin{equation*}
\mathbf{x} \cdot \mathbf{y} = (x_1, x_2) \cdot (y_1, y_2) = x_1 y_1 + x_2 \, y_2
\end{equation*}

Since $\mathbf{x} \perp \mathbf{y}$, we have:

\begin{equation*}
\mathbf{x} \cdot \mathbf{y} = |\mathbf{x}| \, |\mathbf{y}| \, \cos(\phi) = 0
\end{equation*}

which gives:

\begin{equation*}
x_1 y_1 + x_2 \, y_2  = 0, \quad \text{or equivalently} \quad y_1 = x_2, \, y_2 = -x_1
\end{equation*}

Substituting and expanding, Gauss's theorem for a 2D configuration becomes:

\begin{equation*}
\iint\limits_{\Omega} \left( \frac{\partial F_x}{\partial x} + \frac{\partial F_y}{ \partial y} \right) \, \mathrm{d} x \, \mathrm{d} y 
= \int\limits_{\partial \Omega} F_x \, \mathrm{d} y - F_y \, \mathrm{d} x
\end{equation*}

---

**Explanation of variables:**

- $F_x, F_y$: components of the 2D vector field $\mathbf{F}(x,y)$ along $x$ and $y$ axes  
- $\Omega$: area in the $xy$-plane  
- $\partial \Omega$: boundary curve of the area $\Omega$  
- $\mathrm{d} \mathbf{l_\perp}$: vector perpendicular to the line element along the boundary  
- $dx, dy$: differentials along the tangent direction of the curve  
- This shows that the sum of divergence over the area equals the flux through the boundary, consistent with Gauss's theorem in 2D.





## Symbolic Calculation

Symbolic computation of the vector field flux $\mathbf{F}(x,y)$:

\begin{equation*}
\mathbf{F}(x,y)  =   \left( -y \, (1-x-y)+ \frac{x}{4} \right) \, \mathbf{i}   +   \left(2 -x^2 + \frac{y}{4} \right) \, \mathbf{j}  
\end{equation*}

in the 2D plane through a curve defined by the parametric equations:

\begin{align*}
x &= \cos(t) \\
y &= \sin(t)
\end{align*}

After substitution, the expression for computing the flux through the boundary of the 2D region becomes:

\begin{equation*}
\Phi = \int\limits_{\partial \Omega} F_x(x,y) \, \mathrm{d} y - F_y(x,y) \, \mathrm{d} x
= \int\limits_{C} \left( F_x(t) \, \cos(t) - F_y(t) \, (-\sin(t)) \right) \mathrm{d} t
\end{equation*}

---

**Explanation of variables:**

- $\mathbf{F}(x,y) = (F_x, F_y)$: 2D vector field components along $x$ and $y$  
- $\partial \Omega$: boundary curve of the 2D region  
- $x = \cos(t), y = \sin(t)$: parametric equations describing the curve $C$  
- $\mathrm{d}x, \mathrm{d}y$: differentials along the curve  
- $\Phi$: flux of the vector field through the boundary of the region  
- The integral shows that the total flux is computed as the line integral of the vector field along the boundary, consistent with Gauss's theorem in 2D.



```python
%reset -sf 
from utils.utils import *
from sympy import *
from sympy.vector import CoordSys3D, divergence
    
q = CoordSys3D('q')                  # suradnicova sustava
t, r = symbols("t,r")                # parameter pre vypocet integralu
                                     # vektorove pole a substitucia parametrov
F = (-q.y*(1 - q.x - q.y) + q.x/4) * q.i + (2 - q.x**2 + q.y/4) * q.j
Fs = F.subs([(q.x, cos(t)), (q.y, sin(t))])
                                    
ns = (cos(t)*q.i) - (-sin(t))*q.j    # normalovy vektor dl
Ph = integrate(Fs.dot(ns), (t, 0, 2*pi) )

ltxprint('\mathbf{F}(x,y)', F)
ltxprint('\Phi', Ph)
```



## Numerical Calculation

For numerical computation, we use the function *quad()* to calculate the definite integral of a function.


```python
from scipy.integrate import quad
import numpy as np

def Fp(t):
    x = np.cos(t)
    y = np.sin(t)
    return (-y * (1-x-y)+ x/4)*(np.cos(t)) - (2 - x**2 + y/4) * (-np.sin(t))

Flux, err = quad(Fp, 0, 2*np.pi)
print(Flux)
```



For comparison and illustration, we repeat the calculation for the left-hand side of the expression by computing the flux through the area of a circle with radius $r$. If we denote

\begin{equation*}
D(x,y) = \nabla \cdot \mathbf{F}(x,y)
\end{equation*}

then after substitution:

\begin{equation*}
\Phi = \iint\limits_{\Omega}  \nabla \cdot \mathbf{F}(x,y)  \, \mathrm{d} x \, \mathrm{d} y 
= \int\limits_0^{2 \pi} \int\limits_0^1 D( r \cos(\psi), r \sin(\psi) ) \, r \, \mathrm{d} r \, \mathrm{d} \psi
\end{equation*}

---

**Explanation of variables:**

- $D(x,y)$: divergence of the vector field $\mathbf{F}(x,y)$ at point $(x,y)$  
- $r$: radial coordinate in polar coordinates  
- $\psi$: angular coordinate in polar coordinates  
- $\Omega$: circular area of radius 1 in the 2D plane  
- $\Phi$: total flux of the vector field through the area, computed using the area integral in polar coordinates


```python
D = divergence(F)
Ds = D.subs([(q.x, r*cos(t)), (q.y, r*sin(t))])
Ph = integrate(  integrate(Ds*r, (r, 0, 1)), (t, 0, 2*pi) )

ltxprint('\Phi', Ph)
```
