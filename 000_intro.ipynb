# Introduction

The notebook is focused on presenting the possibilities of simulating the propagation of electromagnetic waves in a medium using the FDTD method. The introductory chapters provide a brief overview of vector analysis concepts along with methods for manipulating vector fields in Python. This is followed by a derivation of the FDTD method and elementary one-dimensional examples demonstrating the fundamental principles and properties of the method. For solving more complex problems, the MEEP software environment is described together with usage examples.

The use of this notebook assumes basic knowledge of the Python programming language and elementary knowledge of vector analysis concepts.
