

# Stokes' Theorem

Stokes' theorem relates the line integral of a vector field $\mathbf{F}$ along a closed curve $C$ to the surface integral of the curl of the field over the surface bounded by the curve:

\begin{equation}
\oint\limits_C \mathbf{F} \cdot \mathrm{d} \mathbf{l} = \int\limits_S \left(\nabla \times \mathbf{F}\right) \cdot \mathrm{d}\mathbf{s} 
\end{equation}

**Notes:**

- $C$: closed curve bounding the surface $S$  
- $\mathbf{F}$: vector field  
- $\mathrm{d} \mathbf{l}$: differential element along the curve $C$  
- $S$: surface bounded by $C$  
- $\mathrm{d}\mathbf{s}$: differential oriented area element on the surface  
- $\nabla \times \mathbf{F}$: curl of the vector field  

Stokes' theorem in 2D is known as **Green's theorem**.





## Discrete Approximation of Stokes' Theorem

Assume that in the vector field $\mathbf{F}$ there is a surface $S$ bounded by a curve $C$, which we divide into many small rectangular elements $ABCD$ of size $\mathrm{d} x \times \mathrm{d} y$. 

<img src="./img/stokes_02.png" width=500 alt="Vector field on surface" scale="1.0">

The vector field $\mathbf{F}$ can be formally expressed at the center of a small element located at position $(x,y)$ as a superposition of two perpendicular components:

\begin{equation*}
\mathbf{F} =  F_x(x,y) \, \mathbf{i}  +  F_y(x,y)\, \mathbf{j}   = u(x,y) \, \mathbf{i}  + v(x,y) \, \mathbf{j}  
\end{equation*}

For clarity and simplicity of notation, we introduce the shorthand:

\begin{align*}
u &= u(x,y) \\
v &= v(x,y)
\end{align*}

---

**Explanation of variables:**

- $\mathbf{F}(x,y) = F_x \mathbf{i} + F_y \mathbf{j}$: vector field components along $x$ and $y$  
- $u(x,y), v(x,y)$: alternative notation for $F_x$ and $F_y$  
- $S$: surface bounded by the curve $C$  
- $ABCD$: small rectangular element of the surface with dimensions $\mathrm{d}x \times \mathrm{d}y$  
- $(x,y)$: coordinates of the element center  
- This setup allows discretizing the surface integral in Stokes' theorem using small patches.




We are interested in the value of the flux of the vector field along the curve bounding the small element — in our case, a square. The expression can be split into separate contributions from each side of the element:

\begin{align*}
\mathbf{F} \, \mathrm{d} \mathbf{l}  &=
u \, \big(x, y - \frac{\mathrm{d} y}{2}\big) \, \mathrm{d} x
+  v \, \big(x + \frac{\mathrm{d} x}{2}, y\big) \, \mathrm{d} y
\\ &-  u \, \big(x, y + \frac{\mathrm{d} y}{2}\big) \, \mathrm{d} x
-  v \, \big(x - \frac{\mathrm{d} x}{2}, y\big) \, \mathrm{d} y
\end{align*}

In this decomposition, we use the direction of integration and the orthogonality between the field components and the oriented line element; in the scalar (dot) product, only the parallel components contribute:

\begin{equation*}
\mathbf{F}(x,y) \, \mathrm{d} \mathbf{l} = \big( u(x,y) \, \mathbf{i} + v(x,y) \, \mathbf{j} \big) \cdot \big( \mathbf{i} \, \mathrm{d} x + \mathbf{j} \, \mathrm{d} y \big)
\end{equation*}

The approximate value of the field component at the center of the element edge can be obtained using a Taylor series expansion. For edge $AB$:

\begin{equation*}
u\big(x, y-\frac{\mathrm{d} y}{2}\big) \approx u(x,y) - \frac{\partial u(x,y)}{\partial y} \frac{\mathrm{d} y}{2} = u - \frac{\partial u}{\partial y} \frac{\mathrm{d} y}{2}
\end{equation*}

Similarly, we expand for the other edges of the element.

---

**Explanation of variables:**

- $\mathbf{F}(x,y) = u(x,y) \mathbf{i} + v(x,y) \mathbf{j}$: vector field components  
- $\mathrm{d} \mathbf{l}$: differential element along the curve  
- $u(x, y-\mathrm{d}y/2)$: component of the field at the midpoint of edge $AB$  
- $\partial u / \partial y$: partial derivative of $u$ with respect to $y$  
- $\mathrm{d}x, \mathrm{d}y$: lengths of the edges of the small rectangular element  
- The formula approximates the flux contribution along each edge using the midpoint and first-order Taylor expansion.





By substituting and simplifying, we obtain:

\begin{align*}
\mathbf{F} \, \mathrm{d} \mathbf{l}  &=
\big( u - \frac{\partial u}{\partial y} \frac{\mathrm{d} y}{2} \big) \, \mathrm{d} x
+  \big( v + \frac{\partial v}{\partial x} \frac{\mathrm{d} x}{2} \big) \, \mathrm{d} y
\\ &-  \big( u + \frac{\partial u}{\partial y} \frac{\mathrm{d} y}{2} \big) \, \mathrm{d} x
- \big( v - \frac{\partial v}{\partial x} \frac{\mathrm{d} x}{2} \big) \, \mathrm{d} y
\\ &=  \big( - \frac{\partial u}{\partial y} + \frac{\partial v}{\partial x} \big) \, \mathrm{d} x \, \mathrm{d} y
\end{align*}

---

**Explanation of variables:**

- $u, v$: components of the vector field $\mathbf{F}$ along $x$ and $y$  
- $\partial u / \partial y$, $\partial v / \partial x$: partial derivatives of the components along the perpendicular direction  
- $\mathrm{d}x, \mathrm{d}y$: lengths of the edges of the small rectangular element  
- $\mathbf{F} \, \mathrm{d} \mathbf{l}$: approximate contribution of the vector field along the boundary of the element  
- The final expression $\left(-\frac{\partial u}{\partial y} + \frac{\partial v}{\partial x}\right) \, \mathrm{d} x \, \mathrm{d} y$ corresponds to the **curl** (rotation) component perpendicular to the 2D plane.




The expression $\mathrm{d} x \, \mathrm{d} y$ corresponds to the area of an elementary surface $\mathrm{d} s$. If we now "tile" the entire surface $S$ with such small elements, the flux contributions along the edges of neighboring elements cancel out, leaving only the flux along the outer boundary of the surface. This can be written as:

\begin{equation*}
\oint\limits_C \mathbf{F} \, \mathrm{d} \mathbf{l}  =  \int\limits_C F_x \, \mathrm{d} x + F_y \, \mathrm{d} y  = \int\limits_S \big( -\frac{\partial F_x}{\partial y} + \frac{\partial F_y}{\partial x} \big) \, \mathrm{d} s
\end{equation*}

<img src="./img/stokes_10.png" width=500 alt="Flux along the boundary of the surface" scale="0.4">

By changing the sign of $F_x$ and rearranging, we obtain the familiar form of **Green's theorem**:

\begin{equation}
\int\limits_C F_y \, \mathrm{d} y  - F_x \, \mathrm{d} x = \int\limits_S \big( \frac{\partial F_x}{\partial y} + \frac{\partial F_y}{\partial x} \big) \, \mathrm{d} s
\end{equation}

---

**Explanation of variables:**

- $\mathbf{F} = (F_x, F_y)$: vector field in 2D  
- $C$: closed curve bounding the surface $S$  
- $\mathrm{d} \mathbf{l}$: line element along the curve  
- $\mathrm{d} s = \mathrm{d} x \, \mathrm{d} y$: area element of the surface  
- The left-hand side represents the **circulation** of the vector field along the boundary  
- The right-hand side represents the **curl** of the field over the surface, giving the flux through $S$
