


# The FDTD Yee Algorithm

To solve the system of Maxwell's differential equations in the time domain, mathematician [Kane Yee](https://en.wikipedia.org/wiki/Finite-difference_time-domain_method) introduced an algorithm in 1966 that bears his name. The essence of the algorithm is the calculation of time and space derivatives by approximating the derivative using finite differences.

## Approximation of Spatial and Temporal Derivatives

To describe the electromagnetic field in space and time, we introduce a notation that describes the discrete space-time localization of the **E** and **H** field components:

$$
\begin{align*}
\mathbf{E}(\mathbf{r},t) &\approx \mathbf{E}(m \Delta x, n \Delta y, k \Delta z, q \Delta t) = \mathbf{E}^q[m,n,k] \\
\mathbf{H}(\mathbf{r},t) &\approx \mathbf{H}(m \Delta x, n \Delta y, k \Delta z, q \Delta t) = \mathbf{H}^q[m,n,k] \\
\end{align*}
$$

where $m,n,k$ are the indices of the spatial steps in the direction of the axes, and $q$ is the index of the time step. The derivative of a field component at a certain point in space and time, for example **H** at the point $(m \Delta x, n \Delta y, k \Delta z)$ in the z-axis direction and at time step $q$, can then be formally approximated as:

$$
\frac{\partial \mathbf{H}^q[m,n,k]}{\partial z} \approx \frac{\mathbf{H}^q \left[ m,n,k + \frac{1}{2} \right] - \mathbf{H}^q \left[m, n, k- \frac{1}{2} \right ]}{\Delta z} + O[(\Delta z)^2]
$$

From the approximation (*two-point centered difference method*), it follows that the field components should be defined at integer grid points (m, n, k), as well as between grid points at half-step distances (e.g., k+1/2, k-1/2). The derivative of a specific field component, for example $H_z$ in the z-axis direction, can be approximated as:

$$
\frac{\partial H_z^q [m,n,k]}{\partial z} \approx \frac{ H_z^q \left[ m,n,k + \frac{1}{2} \right] - H_z^q \left[m, n, k- \frac{1}{2} \right ]}{\Delta z} + O[(\Delta z)^2]
$$

**Note** - The components **E** and **H** are vector fields, so each component of the field (e.g., $E_x$, $H_z$) is a scalar function that can depend on all coordinates.

## Yee Algorithm in 1D

Let us assume a one-dimensional space in which an electromagnetic wave can propagate in the direction of the $x$-axis. The electromagnetic wave consists only of field components in the direction of the $y$ and $z$ axes, and these are functions of the coordinate $x$.

According to Faraday's law, we can then write:

$$
-\mu \frac{\partial \mathbf{H}}{\partial t} =
-\mu \left[ \frac{\partial H_y}{\partial t}\mathbf{j} + \frac{\partial H_z}{\partial t}\mathbf{k} \right]
= \nabla \times \mathbf{E} =
\begin{vmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\
{\dfrac{\partial}{\partial x}} & 0 & 0 \\
0 & E_y & E_z \end{vmatrix} =
-\frac{\partial E_z}{\partial x} \mathbf{j}  +  \frac{\partial E_y}{\partial x} \mathbf{k}
$$

And from Ampère's law, it follows that:

$$
\epsilon \frac{\partial \mathbf{E}}{\partial t} =
\epsilon \left[ \frac{\partial E_y}{\partial t} \mathbf{j} + \frac{\partial E_z}{\partial t} \mathbf{k} \right]=
\nabla \times \mathbf{H} =
\begin{vmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\
{\dfrac{\partial}{\partial x}} & 0 & 0 \\
0 & H_y & H_z \end{vmatrix} =
-\frac{\partial H_z}{\partial x}\mathbf{j} + \frac{\partial H_y}{\partial x}\mathbf{k}
$$

### Decoupling into Scalar Equations

By comparing the vector components on both sides of the equations above, we can separate the single vector equations into two independent sets of scalar equations. Each set describes a wave with a specific polarization.

#### Set 1: ($E_z, H_y$) Polarization

By equating the coefficients of the **j** unit vector from Faraday's and Ampère's laws, we get one set of coupled equations:

$$ \mu \frac{\partial H_y}{\partial t} = \frac{\partial E_z}{\partial x} $$
$$ \epsilon \frac{\partial E_z}{\partial t} = \frac{\partial H_y}{\partial x} $$

These two equations describe a wave where the electric field oscillates only in the z-direction, and the magnetic field oscillates only in the y-direction. This wave propagates along the x-axis.

#### Set 2: ($E_y, H_z$) Polarization

By equating the coefficients of the **k** unit vector, we get a second, independent set of equations:

$$ -\mu \frac{\partial H_z}{\partial t} = \frac{\partial E_y}{\partial x} $$
$$ \epsilon \frac{\partial E_y}{\partial t} = -\frac{\partial H_z}{\partial x} $$

These equations describe a different wave, where the electric field oscillates in the y-direction and the magnetic field oscillates in the z-direction, also propagating along the x-axis.

In a 1D FDTD simulation, we typically solve only one of these sets of scalar equations, as they are completely independent of each other.

### Explanation of Symbols

* **$\mu$**: The **permeability** of the medium, a measure of its ability to support the formation of a magnetic field.
* **$\epsilon$**: The **permittivity** of the medium, a measure of how an electric field affects and is affected by the medium.
* **$\mathbf{E}$**: The electric field vector.
* **$\mathbf{H}$**: The magnetic field intensity vector.
* **$E_y, E_z$**: The scalar components of the electric field in the y and z directions.
* **$H_y, H_z$**: The scalar components of the magnetic field in the y and z directions.
* **$t$**: Time.
* **$x$**: The spatial coordinate along which the wave propagates.
* **$\mathbf{i}, \mathbf{j}, \mathbf{k}$**: The unit vectors for the x, y, and z axes, respectively.
* **$\frac{\partial}{\partial t}, \frac{\partial}{\partial x}$**: Partial derivative operators with respect to time and space.
* **$\nabla \times$**: The curl operator, which describes the circulation of a vector field.
