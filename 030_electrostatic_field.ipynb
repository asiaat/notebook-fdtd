## Coulomb's Law

Coulomb's law describes the force interaction between charges.  

<img src="./img/obr_f12.png" width=350 alt="Force between two charges" scale="0.3">

A charge $Q_2$ exerts in vacuum a force $\mathbf{F}_{1}$ on charge $Q_1$ at a distance $r$, which can be expressed in vector form as

\begin{equation}
\mathbf{F}_{1} = \frac{1}{4 \pi \epsilon_0} \frac{Q_1 \, Q_2}{{r^2}} \, \mathbf{r}_{21}
\end{equation}

where $\mathbf{r}_{21}$ is a unit vector pointing from $Q_2$ to $Q_1$.  
The vector form automatically accounts for the polarity of the charges: like charges repel, opposite charges attract.  
According to Newton’s third law, we have $\mathbf{F}_{1} = -\mathbf{F}_{2}$.

In scalar form:

\begin{equation}
| \mathbf{F}| = \frac{1}{4 \pi \epsilon_0} \frac{ | Q_1 \, Q_2 |}{{r^2}}
\end{equation}




## Electric Field Intensity

The force acting on a charge $Q_1$ can be expressed using the electric field $\mathbf{E}$ created by charge $Q_2$ in its surroundings:

\begin{equation}
\mathbf{F} = Q_1 \left( \frac{1}{4 \pi \epsilon_0} \frac{Q_2}{{r^2}}  \mathbf{r}_{21} \right) = Q_1 \, \mathbf{E}
\end{equation}

Every static charge creates an electric field in its vicinity:

\begin{equation}
\mathbf{E} = \frac{1}{4 \pi \epsilon_0} \frac{Q}{{r^2}} \, \frac{\mathbf{r}}{r} = \frac{1}{4 \pi \epsilon_0} \frac{Q}{{r^2}} \, \mathbf{r}_0
\end{equation}

where $\mathbf{r}_0$ is a unit vector.  
The direction of the electric field vector is defined as pointing from a positive charge toward a negative charge — this convention originates from historical reasons.


```python
%reset -f 
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

dx = np.linspace(-1.0, 1.0, 10)             # coordinate ranges
dy = np.linspace(-1.0, 1.0, 10)
x, y = np.meshgrid(dx, dy)

[Ex, Ey] = [ 1.0/(x**2+y**2)*x,  1.0/(x**2+y**2)*y ]  # components of the vector field

fig = plt.figure(figsize=(4, 4))
ax = fig.add_subplot(111)
ax.quiver(dx, dy, Ex, Ey, color='r')
plt.xlabel('x')
plt.ylabel('y')
plt.axis('equal')
plt.grid()
plt.show()
```




## Electric Potential

To move a charge $q$ between two points $a$ and $b$ in the electric field of another charge $Q$, work must be performed, which is defined as

\begin{equation}
A = \int\limits_a^b \mathbf{F} \, \mathrm{d} \mathbf{r} = \int\limits_a^b q \, \mathbf{E} \, \mathrm{d} \mathbf{r}
\end{equation}

Assuming $q$ is a unit charge, this becomes

\begin{equation}
A = \int\limits_a^b \mathbf{E} \, \mathrm{d} \mathbf{r} 
= \frac{1}{4 \pi \epsilon_0} \int\limits_a^b \frac{Q}{r^2} \, \mathrm{d} r 
= \frac{Q}{4 \pi \epsilon_0} \left[ -\frac{1}{r} \right]_a^b 
= \frac{Q}{4 \pi \epsilon_0} \left( -\frac{1}{b} + \frac{1}{a} \right)
\end{equation}

The work required to move a unit charge from point $a$ to point $b$ is therefore equal to the difference between the work to bring the charge to $a$ and then to $b$. Since the initial value at point $a$ is unknown, it is convenient to define a reference point with zero potential. Because the work to move a charge in the electric field of another charge is inversely proportional to the distance, we choose infinity as the reference point with zero potential.

Thus, the work required to move a unit charge from infinity to a distance $r$ from charge $Q$ is

\begin{equation}
A = - \frac{Q}{4 \pi \epsilon_0} \frac{1}{r} 
= - \frac{Q}{4 \pi \epsilon_0} \frac{1}{r^2} \, \mathbf{r}_0 \cdot \mathbf{r} 
= - \mathbf{E} \cdot \mathbf{r}
\end{equation}

### Variables

- $q$ — the test charge being moved (here assumed as 1 for unit charge)  
- $Q$ — source charge creating the electric field  
- $\epsilon_0$ — vacuum permittivity  
- $r$ — distance between the test charge and source charge  
- $\mathbf{r}$ — position vector from the source charge to the test charge  
- $\mathbf{r}_0$ — unit vector in the direction of $\mathbf{r}$  
- $\mathbf{E}$ — electric field at the position of the test charge  
- $A$ — work required to move the charge




## Electric Potential at a Point

The potential at a point located at distance $r$ from a point charge, which creates an electric field $\mathbf{E}$ in its surroundings, is defined as

\begin{equation}
V(r) = -A =  \mathbf{E} \cdot \mathbf{r}
\end{equation}

**TODO - IMAGE**

The potential difference between two points $a$ and $b$ can be determined as

\begin{equation}
V_{ab} = \int\limits_a^b \mathbf{E} \cdot \mathrm{d} \mathbf{r} = V_b - V_a
\end{equation}

Since the electric field is conservative, the value of the line integral depends only on its endpoints. Therefore, the integral over a closed curve is zero (the start and end points coincide):

\begin{equation}
\oint\limits_C \mathbf{E} \cdot \mathrm{d} \mathbf{r} = 0
\end{equation}

Using Stokes' theorem, this relationship can be rewritten as

\begin{equation}
\oint\limits_C \mathbf{E} \cdot \mathrm{d} \mathbf{r} = \int\limits_S \big( \nabla \times \mathbf{E} \big) \cdot \mathrm{d}\mathbf{s} = 0
\end{equation}

Or in differential form:

\begin{equation}
\nabla \times \mathbf{E} = 0
\end{equation}

This states that an electrostatic field does not form loops.

### Variables

- $V(r)$ — electric potential at distance $r$ from the source charge  
- $A$ — work done to bring a unit charge to the point  
- $\mathbf{E}$ — electric field vector  
- $\mathbf{r}$ — position vector from the source charge to the point  
- $V_{ab}$ — potential difference between points $a$ and $b$  
- $C$ — closed curve in space  
- $S$ — surface bounded by curve $C$




## Gauss's Law

Let's calculate the flux of an electric field through a closed, smooth surface that encloses a point charge. For simplicity, choose a spherical surface with the charge at its center. In spherical coordinates, the flux of the field is then

\begin{equation*}
\oint\limits_S \mathbf{E} \, \mathrm{d} \mathbf{s} 
= \int\limits_{\phi = 0}^{\pi} \int\limits_{\zeta = 0}^{2 \pi} \frac{Q}{4 \pi \epsilon_0} \frac{1}{r^2} \, r^2 \sin(\phi) \, \mathrm{d} \phi \, \mathrm{d} \zeta 
= \frac{Q}{\epsilon_0}
\end{equation*}

In general, the result does not depend on the shape of the surface (as long as it is smooth and continuous), nor on whether the charge is a point at the center or distributed within the volume enclosed by the surface. For a volume-distributed charge with density $\rho$ in volume $V$, the relation, known as Gauss's law in integral form, is

\begin{equation}
\oint\limits_S \mathbf{E} \, \mathrm{d} \mathbf{s} = \frac{1}{\epsilon_0} \int\limits_V \rho(r) \, \mathrm{d}v
\end{equation}

Using the divergence theorem (Gauss's theorem), we get

\begin{equation*}
\oint\limits_S \mathbf{E} \, \mathrm{d} \mathbf{s} = \int\limits_V \nabla \cdot \mathbf{E} \, \mathrm{d}v = \frac{1}{\epsilon_0} \int\limits_V \rho(r) \,  \mathrm{d}v
\end{equation*}

Comparing the integrands, we obtain Gauss's law in differential form, which defines the divergence of the electric field at a given point in space:

\begin{equation}
\nabla \cdot \mathbf{E} = \frac{\rho(r)}{\epsilon_0}
\end{equation}

### Variables

- $\mathbf{E}$ — electric field vector  
- $\mathbf{s}$ — infinitesimal oriented surface element  
- $S$ — closed surface enclosing the charge(s)  
- $Q$ — point charge at the center of the surface  
- $\epsilon_0$ — vacuum permittivity  
- $\rho(r)$ — charge density at position $r$  
- $V$ — volume enclosed by surface $S$  
- $\nabla \cdot \mathbf{E}$ — divergence of the electric field




### Example – Electric Field of a Charged Sphere

Suppose we have a conductive sphere with radius $r$ that is charged with a charge $Q$. From Gauss's law, we have

\begin{equation*}
\oint\limits_S \mathbf{E} \, d \mathbf{s} = \frac{Q}{\epsilon_0}
\end{equation*}

Assuming that the magnitude of the field on the surface of the sphere is constant, we can rewrite the relation using the surface area of a sphere:

\begin{equation*}
E(r) \, 4 \pi r^2 = \frac{Q}{\epsilon_0}
\end{equation*}

For the electric field on the surface of the charged sphere, we then have

\begin{equation*}
\mathbf{E} = \frac{Q}{4 \pi \epsilon_0 r^2} \mathbf{r}_0
\end{equation*}

Comparing with the expression for the field of a point charge, it follows that on the surface of a conductive sphere the field behaves as if all the charge were concentrated at the center. The potential of the charged sphere is then

\begin{equation*}
V(r) = \mathbf{E} \cdot \mathbf{r} = \frac{Q}{4 \pi \epsilon_0 r^2} \mathbf{r}_0 \mathbf{r} = \frac{Q}{4 \pi \epsilon_0 r}
\end{equation*}

### Variables

- $\mathbf{E}$ — electric field vector  
- $r$ — radius or distance from the center of the sphere  
- $Q$ — total charge on the sphere  
- $\epsilon_0$ — vacuum permittivity  
- $\mathbf{r}_0$ — unit vector in the radial direction  
- $V(r)$ — electric potential at distance $r$





## Capacitance

Capacitance quantifies the ability of a body to store electric charge. It is defined as the ratio of the charge on the body to its potential. The potential can be measured relative to a reference zero potential at infinity, which is called **self-capacitance**, or relative to another body, which is called **mutual capacitance**.

\begin{equation}
C = \frac{Q}{V}
\end{equation}

**Example – Self-Capacitance of the Earth**

Suppose we have a sphere in free space. Substituting the potential of the sphere into the capacitance formula, we get

\begin{equation*}
C = \frac{Q}{\frac{Q}{4 \pi \epsilon_0 r}} = 4 \pi \epsilon_0 \, r
\end{equation*}

### Variables

- $C$ — capacitance  
- $Q$ — charge on the body  
- $V$ — electric potential of the body  
- $r$ — radius of the sphere  
- $\epsilon_0$ — vacuum permittivity



```python
# Kapacita zemegule
from scipy import pi
from scipy.constants import epsilon_0    # import konstant z databazy

r = 6370e3                                # polomer zeme [m]
Cap = 4*pi*epsilon_0*r
print('C= {0:.3f} [uF]'.format(Cap*1e6) )
```
