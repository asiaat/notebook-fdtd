

# Curl (Rotation)

The vector cross product of the operator $\nabla$ with a vector field gives a vector field that describes the **rotation** or **vorticity** of the vector field, i.e., how much the field is swirling or rotating at a given point in space.  

The curl vector at a point points in the direction of the **axis of rotation**, and its orientation relative to the plane of rotation is determined using the **right-hand rule**.

\begin{equation}
\mathrm{curl} \, \mathbf{F} = \nabla \times \mathbf{F} = 
\begin{vmatrix} 
\mathbf{i} & \mathbf{j} & \mathbf{k} \\
{\dfrac{\partial}{\partial x}} & {\dfrac{\partial}{\partial y}} & {\dfrac{\partial}{\partial z}} \\
F_x & F_y & F_z 
\end{vmatrix} =
\left(\frac{\partial F_z}{\partial y} - \frac{\partial F_y}{\partial z}\right) \mathbf{i} +
\left(\frac{\partial F_x}{\partial z} - \frac{\partial F_z}{\partial x} \right) \mathbf{j} +
\left(\frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y} \right) \mathbf{k} 
\end{equation}

**Explanation of variables:**

- $\mathbf{F} = (F_x, F_y, F_z)$: the vector field, with components along the $x$, $y$, and $z$ axes.  
- $F_x, F_y, F_z$: scalar functions representing the vector field components.  
- $\mathbf{i}, \mathbf{j}, \mathbf{k}$: unit vectors along the $x$, $y$, and $z$ directions.  
- $\nabla \times \mathbf{F}$: the **curl** of $\mathbf{F}$, producing a vector field whose magnitude represents the local rotational strength and whose direction represents the axis of rotation.  

**Additional Notes:**  
- The right-hand rule determines the direction of the curl vector: if the fingers of your right hand follow the rotation in the plane, your thumb points in the direction of the curl.  
- Curl is widely used in **fluid dynamics** (vorticity), **electromagnetism** (magnetic fields), and other vector field analyses.




## Curl in Cylindrical Coordinates

In cylindrical coordinates $(r, \varphi, z)$, the curl of a vector field $\mathbf{F}$ is given by:

\begin{equation}
\nabla \times \mathbf{F} =
\left(\frac{1}{r}\frac{\partial F_z}{\partial \varphi} - \frac{\partial F_\varphi}{\partial z}\right) \mathbf{r} +
\left(\frac{\partial F_r}{\partial z} - \frac{\partial F_z}{\partial r}\right) \mathbf{\varphi} +
\frac{1}{r}\left(\frac{\partial ( r F_\varphi )}{\partial r} - \frac{\partial F_r}{\partial \varphi}\right) \mathbf{z}
\end{equation}

**Explanation of variables:**

- $\mathbf{F} = (F_r, F_\varphi, F_z)$: the vector field in cylindrical coordinates.  
- $F_r$: radial component of the vector field along $\mathbf{r}$.  
- $F_\varphi$: angular (azimuthal) component along $\mathbf{\varphi}$.  
- $F_z$: vertical component along $\mathbf{z}$.  
- $\mathbf{r}, \mathbf{\varphi}, \mathbf{z}$: unit vectors in the radial, angular, and vertical directions, respectively.  
- $\nabla \times \mathbf{F}$: the curl of $\mathbf{F}$, representing the **local rotational tendency** of the vector field in 3D cylindrical coordinates.  

**Notes:**  
- The factor $1/r$ appears in terms involving angular derivatives to account for the geometry of cylindrical coordinates.  
- The direction of the curl vector is determined using the **right-hand rule**, similar to Cartesian coordinates.

## Curl in Spherical Coordinates

In spherical coordinates $(r, \theta, \varphi)$, the curl of a vector field $\mathbf{F}$ is given by:

\begin{equation}
\nabla \times \mathbf{F} =
\frac{1}{r\sin\theta}\left(\frac{\partial}{\partial \theta} ( F_\varphi \sin\theta )
    - \frac{\partial F_\theta}{\partial \varphi}\right) \mathbf{r} +
\frac{1}{r}\left(\frac{1}{\sin\theta}\frac{\partial F_r}{\partial \varphi} 
    - \frac{\partial}{\partial r} ( r F_\varphi ) \right) \mathbf{\theta} +
\frac{1}{r}\left(\frac{\partial}{\partial r} ( r F_\theta )
    - \frac{\partial F_r}{\partial \theta}\right) \mathbf{\varphi}
\end{equation}

**Explanation of variables:**

- $\mathbf{F} = (F_r, F_\theta, F_\varphi)$: the vector field in spherical coordinates.  
- $F_r$: radial component along $\mathbf{r}$.  
- $F_\theta$: polar (colatitude) component along $\mathbf{\theta}$.  
- $F_\varphi$: azimuthal component along $\mathbf{\varphi}$.  
- $\mathbf{r}, \mathbf{\theta}, \mathbf{\varphi}$: unit vectors in the radial, polar, and azimuthal directions, respectively.  
- $\nabla \times \mathbf{F}$: the curl of $\mathbf{F}$, representing the local **rotation** of the vector field in 3D spherical coordinates.  

**Notes:**  
- The factors $1/r$ and $1/(r\sin\theta)$ account for the geometry of spherical coordinates.  
- The direction of the curl vector follows the **right-hand rule**, similar to Cartesian and cylindrical coordinates.




## Properties of Curl (Rotation)

If $\mathbf{F}, \mathbf{G}$ are vector fields, $f$ is a scalar field, and $a, b$ are real numbers, then the curl operator satisfies the following identities:

- **Linearity of curl with respect to real numbers:**

\begin{equation*}
\nabla \times \left(a \, \mathbf{F}+b \, \mathbf{G} \right) = 
a \, \nabla \times \mathbf{F} + b \, \nabla \times \mathbf{G}
\end{equation*}

- **Curl of a gradient is zero:**

\begin{equation*}
\nabla \times \nabla f = \mathrm{curl} \, \mathrm{grad} \, f = \mathbf{0}
\end{equation*}

- **Curl of a vector field multiplied by a scalar function:**

\begin{equation*}
\nabla \times \left(f \mathbf{F}\right) = 
\nabla f \times \mathbf{F} + f \, \nabla \times \mathbf{F}
\end{equation*}

- **Curl of the cross product of two vector fields:**

\begin{equation*}
\nabla \times \left(\mathbf{F} \times \mathbf{G}\right) = 
\left(\mathbf{G} \cdot \nabla\right)\mathbf{F} - \left(\mathbf{F} \cdot \nabla\right)\mathbf{G}
+ \mathbf{F}\left(\nabla \cdot \mathbf{G}\right) - \mathbf{G}\left(\nabla \cdot \mathbf{F}\right)
\end{equation*}

- **Curl of a curl of a vector field:**

\begin{equation*}
\nabla \times \left(\nabla \times \mathbf{F}\right) =
\nabla\left(\nabla \cdot \mathbf{F}\right) - \Delta \mathbf{F}
\end{equation*}

---

**Explanation of variables:**

- $\mathbf{F}, \mathbf{G}$: vector fields  
- $f$: scalar field  
- $a, b$: real numbers (constants)  
- $\nabla \times$: curl operator  
- $\nabla \cdot$: divergence operator  
- $\nabla f$: gradient of a scalar field  
- $\Delta \mathbf{F}$: vector Laplacian of $\mathbf{F}$  
- $\mathbf{0}$: zero vector  

These identities are fundamental in **vector calculus**, and are widely used in **fluid dynamics**, **electromagnetism**, and other fields involving vector fields.




## Discrete Approximation of Curl in 2D

The curl in the $(x,y)$ plane has only a component in the $z$-direction.  

<img src="./img/rot_2d_01.png" width=400px alt="Curl in the x-y plane" scale="1.25"/>

For small cell sizes, we can approximate the curl as:

\begin{equation*}
\frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y} \approx
\frac{F_y\left(x +\frac{\Delta x}{2}, y\right) - F_y\left(x -\frac{\Delta x}{2}, y\right) }{\Delta x} -
\frac{F_x\left(x, y +\frac{\Delta y}{2} \right) - F_x\left(x, y -\frac{\Delta y}{2}\right) }{\Delta y} 
\end{equation*}

If $\Delta x = \Delta y = \delta$, the formula simplifies to:

\begin{equation}
\frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y} \approx
\frac{1}{\delta} \Big(
F_y\left(x +\frac{\delta}{2}, y\right) - F_y\left(x -\frac{\delta}{2}, y\right) 
- F_x\left(x, y +\frac{\delta}{2} \right) + F_x\left(x, y -\frac{\delta}{2}\right) \Big)
\end{equation}

---

**Explanation of variables:**

- $F_x, F_y$: components of the vector field $\mathbf{F}(x,y)$ in the $x$ and $y$ directions  
- $\Delta x, \Delta y$: small increments in $x$ and $y$ directions for the finite difference approximation  
- $\delta$: uniform grid spacing if $\Delta x = \Delta y$  
- The curl in 2D is oriented along the $z$-axis, perpendicular to the $(x,y)$ plane.  

This is the **finite-difference approximation** of the curl used in numerical simulations of 2D vector fields.




## Symbolic Computation

Using symbolic manipulation with *sympy*, we can determine the curl of a vector field in the 2D plane.  

Consider the vector field:

\begin{equation*}
\mathbf{F}(x,y) = \left( -y \, (1-x-y)+ \frac{x}{4} \right) \mathbf{i} + \left(2 -x^2 + \frac{y}{4} \right) \mathbf{j}
\end{equation*}

**Explanation of variables:**

- $\mathbf{F}(x,y)$: the 2D vector field  
- $F_x = -y \, (1-x-y)+ \frac{x}{4}$: $x$-component of the vector field  
- $F_y = 2 - x^2 + \frac{y}{4}$: $y$-component of the vector field  
- $\mathbf{i}, \mathbf{j}$: unit vectors in the $x$ and $y$ directions  
- The **curl** in 2D is a scalar field representing the $z$-component of rotation perpendicular to the $(x,y)$ plane:

\begin{equation*}
\mathrm{curl}\, \mathbf{F} = \frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y}
\end{equation*}

This can be computed symbolically in *sympy* and then evaluated at specific points or visualized.


```python
%reset -f 
from utils.utils import *
from sympy.vector import CoordSys3D, curl
r = CoordSys3D('R')
F = (-r.y*(1 - r.x - r.y**2) + r.x/4)*r.i + (2 - r.x**2 + r.y/4)*r.j
C = curl(F)

ltxprint(r'\nabla \times \mathbf{F}', C)
```




The value of the curl at a chosen point in the plane $P(x,y) = (-1.5, 0.5)$.



```python
P = [-1.5, 0.5]
value = C.subs([ (r.x, P[0]), (r.y, P[1]) ])
ltxprint(r'\mathrm{curl}(P)', value)
```



Similar to divergence, for visualizing the result of a symbolic computation we can use a suitable substitution of variables in the string (e.g., `r.x` for `x`) and the `eval()` function to evaluate the string.


```python
%matplotlib inline
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

Fx = F.components[r.i].subs([(r.x, 'x'), (r.y, 'y')])
Fy = F.components[r.j].subs([(r.x, 'x'), (r.y, 'y')]) 
Cs = C.subs([(r.x, 'x'), (r.y, 'y'), (r.k, 1)]) 

dx = np.linspace(-2.0, 2.0, 15) 
dy = np.linspace(-2.0, 2.0, 15) 
x,y = np.meshgrid(dx, dy)  

c = eval(str(Cs))
fx = eval(str(Fx))
fy = eval(str(Fy))

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111)
img = ax.imshow(c, extent=[-2, 2, -2, 2], origin='lower', interpolation='kaiser', 
                 clip_on=True, alpha=0.5, cmap='jet') 
fig.colorbar(img)
ax.quiver(x, y, fx, fy, color='r')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```


## Numerical Computation

Using the *numdifftools* library, we can compute the curl of the vector field $\mathbf{F}(x,y)$ from the previous example at the point in the plane $P(x,y) = (-1.5, 0.5)$.  

Numerical libraries like *numpy* and *numdifftools* do not have direct support for computing curl, so we use the **Jacobian matrix** of first partial derivatives:

\begin{equation*}
\mathbf{J} = \begin{bmatrix}
    \dfrac{\partial \mathbf{F}}{\partial x_1} & \cdots & \dfrac{\partial \mathbf{F}}{\partial x_n} 
\end{bmatrix}
= \begin{bmatrix}
    \dfrac{\partial F_1}{\partial x_1} & \cdots & \dfrac{\partial F_1}{\partial x_n}\\
    \vdots & \ddots & \vdots\\
    \dfrac{\partial F_m}{\partial x_1} & \cdots & \dfrac{\partial F_m}{\partial x_n} 
\end{bmatrix}
\end{equation*}

In our 2D case, the Jacobian matrix has the form:

\begin{equation*}
\mathbf{J} 
= \begin{bmatrix}
    \dfrac{\partial F_x}{\partial x} &  \dfrac{\partial F_x}{\partial y}\\
    \dfrac{\partial F_y}{\partial x} &  \dfrac{\partial F_y}{\partial y} 
\end{bmatrix}
\end{equation*}

---

**Explanation of variables:**

- $\mathbf{F} = (F_x, F_y)$: the 2D vector field  
- $F_x, F_y$: components of $\mathbf{F}$ along $x$ and $y$  
- $x_1, \dots, x_n$: general coordinates (here $x_1 = x$, $x_2 = y$)  
- $F_1, \dots, F_m$: general components of $\mathbf{F}$ (here $F_1 = F_x$, $F_2 = F_y$)  
- $\mathbf{J}$: Jacobian matrix containing all first-order partial derivatives, used to approximate the curl numerically in 2D.  





```python
%reset -f 
import numdifftools as nd
import numpy as np

def F(p):
    [x,y] = p
    return np.array([ (-y*(1 - x - y**2) + x/4),  (2 - x**2 + y/4) ])

P = [-1.5, 0.5]
J = nd.Jacobian(F)
value = J(P)
print(value)
```



The value of the curl can be determined directly using the elements of the Jacobian matrix:

\begin{equation*}
\mathrm{curl} \, \mathbf{F} = \frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y} 
\end{equation*}

**Explanation of variables:**

- $F_x, F_y$: components of the 2D vector field $\mathbf{F}(x,y)$ along the $x$ and $y$ directions  
- $\frac{\partial F_y}{\partial x}$: partial derivative of $F_y$ with respect to $x$  
- $\frac{\partial F_x}{\partial y}$: partial derivative of $F_x$ with respect to $y$  
- $\mathrm{curl}\, \mathbf{F}$: scalar value representing the rotation of the vector field in the plane, oriented along the $z$-axis


```python
curl = value[1,0] - value[0,1]
print(curl)
```
